**PLATFORM INSTAGRAM**

**Membaca data mentah dari file scrapping**

```python
import pandas as pd
import os

# 1. Memastikan folder data/processed tersedia
os.makedirs("data/processed", exist_ok=True)

# 2. Membaca data mentah dari folder data/raw/
df = pd.read_csv("/content/drive/MyDrive/magang-ba-2026/data/raw/sd_mv92q98526jbuxxezd(batch2).csv").copy()
```

**Membersihkan data kosong**

```python
# 3. Bersihkan Data Kosong
df = df.dropna(subset=['comment'])
```

**Membuat ID Post**

```python
# 4. Mendapatkan URL unik untuk membuat ID Post
unique_urls = df['input_url'].unique()
post_id_map = {url: f"IG-P{str(i+1).zfill(3)}" for i, url in enumerate(unique_urls)}
df['post_id'] = df['input_url'].map(post_id_map)
```

**Menyiapkan kolom**

```python
# 5. Menyiapkan kolom yang dibutuhkan (tanpa username)
df['text_content'] = df['comment']
df['published_at'] = pd.to_datetime(df['comment_date']).dt.strftime('%Y-%m-%d')
df['eligibility_status'] = 'layak'
```

**Membuat komentar per post ID**

```python
# 6. Membuat ID Komentar berurutan per Post ID
df['comment_num'] = df.groupby('post_id').cumcount() + 1
df['comment_id'] = df['post_id'] + '-C' + df['comment_num'].astype(str).str.zfill(3)
```

```python
# 7. Membatasi maksimal 150 komentar per unggahan
df = df[df['comment_num'] <= 150]
```

```python
# 8. Memilih dan menyusun ulang kolom akhir
final_df = df[['comment_id', 'post_id', 'text_content', 'published_at', 'eligibility_status']]
```

**Menyimpan hasil akhir**

```python
# 9. Menyimpan hasil CSV bersih ke folder data/processed/
final_df.to_csv("data/processed/comments_instagram_b2.csv", index=False)

print("Proses selesai! File bersih berhasil disimpan di: data/processed/comments_instagram_b2.csv")
print(f"Total baris saat ini: {len(final_df)}\n")
print(final_df.head())
```

**Membuat tabel sebaran komentar layak per platform × bulan**

```python
import pandas as pd
import os

# 1. Muat file data yang sudah bersih
file_path = "data/processed/comments_batch2_merged.csv"
df = pd.read_csv(file_path)

# Pastikan kolom tanggal bertipe datetime
df["tgl"] = pd.to_datetime(df['published_at'], errors="coerce")
```

```python
# 2. Filter hanya data dengan status 'layak'
if 'eligibility_status' in df.columns:
    df_layak = df[df['eligibility_status'] == 'layak'].copy()
else:
    df_layak = df.copy()
```

```python
# 3. Ambil ekstrak bulan (format YYYY-MM atau nama bulan)
df_layak['bulan'] = df_layak['tgl'].dt.to_period('M')
```

```python
# 4. Pastikan kolom platform terdeteksi
if 'platform' not in df_layak.columns:
    df_layak['platform'] = df_layak['post_id'].apply(lambda x: 'Instagram' if str(x).startswith('IG') else 'TikTok')
```

```python
# 5. Buat tabel crosstab sebaran platform x bulan
tabel_sebaran = pd.crosstab(df_layak['platform'], df_layak['bulan'])

print("=" * 50)
print("TABEL SEBARAN KOMENTAR LAYAK (Platform × Bulan):")
print("=" * 50)
print(tabel_sebaran)
```

```python
# 6. Simpan ke folder docs/ dengan nama D07_sebaran.csv
os.makedirs("docs", exist_ok=True)
tabel_sebaran.to_csv("docs/D07_sebaran.csv")
print("\nBerhasil menyimpan tabel sebaran ke: docs/D07_sebaran.csv")
```

**Pemeriksaan Mutu**

```python
import pandas as pd
import os

# 1. Muat file data yang sudah bersih
file_path = "data/processed/comments_batch2_merged.csv"
df = pd.read_csv(file_path)

# Pastikan kolom tanggal bertipe datetime & filter rentang valid
df["tgl"] = pd.to_datetime(df['published_at'], errors="coerce")
df_bersih = df[
    df['tgl'].notna() &
    (df['tgl'] >= "2026-07-01") &
    (df['tgl'] <= "2026-09-30 23:59")
].copy()
```

```python
# 2. Lakukan 3 tahap pemeriksaan mutu (QC)
# Aturan: 5 baris acak per 100 komentar (karena ada 202 data, diambil 10 baris per tahap)
qc_samples = []
for i in range(3):
    # Mengambil sampel acak sebanyak 10 baris per tahap (5 baris x 2 ratus data)
    sample = df_bersih.sample(n=10, random_state=42 + i)
    sample['tahap_qc'] = f"Tahap_{i+1}"
    qc_samples.append(sample)

# Gabungkan hasil dari 3 tahap QC
df_qc_final = pd.concat(qc_samples)

print("=" * 50)
print("HASIL PEMERIKSAAN MUTU (QUALITY CONTROL):")
print("=" * 50)
print(f"Total sampel QC terkumpul dari 3 tahap : {len(df_qc_final)} baris")
print(df_qc_final[['tahap_qc', 'post_id', 'published_at']].head(15))
```

```python
# 3. Simpan ke folder docs/ dengan nama D07_sampel_qc.csv
os.makedirs("docs", exist_ok=True)
df_qc_final.to_csv("docs/D07_sampel_qc.csv", index=False)
print("\nBerhasil menyimpan file QC ke: docs/D07_sampel_qc.csv")
```
