**PLATFORM INSTAGRAM**

**Membaca data mentah dari file scrapping**

```python
import pandas as pd
import os

# 1. Memastikan folder data/processed tersedia
os.makedirs("data/processed", exist_ok=True)

# 2. Membaca data mentah dari folder data/raw/
df = pd.read_csv("/content/drive/MyDrive/magang-ba-2026/data/raw/sd_mvagmq2g124cnxjhnt(batch3).csv").copy()
```

**Membersihkan data kosong**

```python
# 3. Bersihkan Data Kosong
df = df.dropna(subset=['comment'])
```

**Membuat ID Post**

```python
# 4. Mendapatkan URL unik untuk membuat ID Post
unique_urls = df['input_url'].unique()
post_id_map = {url: f"IG-P{str(i+1).zfill(3)}" for i, url in enumerate(unique_urls)}
df['post_id'] = df['input_url'].map(post_id_map)
```

**Menyiapkan kolom**

```python
# 5. Menyiapkan kolom yang dibutuhkan (tanpa username)
df['text_content'] = df['comment']
df['published_at'] = pd.to_datetime(df['comment_date']).dt.strftime('%Y-%m-%d')
df['eligibility_status'] = 'layak'
```

**Membuat komentar per post ID**

```python
# 6. Membuat ID Komentar berurutan per Post ID
df['comment_num'] = df.groupby('post_id').cumcount() + 1
df['comment_id'] = df['post_id'] + '-C' + df['comment_num'].astype(str).str.zfill(3)
```

```python
# 7. Membatasi maksimal 150 komentar per unggahan
df = df[df['comment_num'] <= 150]
```

```python
# 8. Memilih dan menyusun ulang kolom akhir
final_df = df[['comment_id', 'post_id', 'text_content', 'published_at', 'eligibility_status']]
```

**Menyimpan hasil akhir**

```python
# 9. Menyimpan hasil CSV bersih ke folder data/processed/
final_df.to_csv("data/processed/comments_instagram_b3.csv", index=False)

print("Proses selesai! File bersih berhasil disimpan di: data/processed/comments_instagram_b3.csv")
print(f"Total baris saat ini: {len(final_df)}\n")
print(final_df.head())
```

**Fungsi normalize_text dan Hitung text_hash**

**Membuat file text_utils.py**

```python
import os

# 1. Buat folder src jika belum ada
os.makedirs("/content/drive/MyDrive/magang-ba-2026/src", exist_ok=True)

# 2. Isi file src/text_utils.py sesuai panduan
text_utils_code = '''import hashlib
import re

def normalize(t: str) -> str:
    t = re.sub(r"\\s+", " ", str(t).lower()).strip()
    return t.strip(".,!?;:")

def text_hash(t: str) -> str:
    return hashlib.sha256(normalize(t).encode("utf-8")).hexdigest()
'''

with open("/content/drive/MyDrive/magang-ba-2026/src/text_utils.py", "w", encoding="utf-8") as f:
    f.write(text_utils_code)

print("File src/text_utils.py berhasil dibuat!")
```

```python
import pandas as pd
import os
import pandas as pd
import hashlib
import re
import sys

# Menambahkan path root proyek ke system path Python agar folder 'src' terbaca
project_path = "/content/drive/MyDrive/magang-ba-2026"
if project_path not in sys.path:
    sys.path.append(project_path)

# Pastikan kita berada di direktori project
os.chdir(project_path)

# Mengimpor dari modul src yang baru kita buat
from src.text_utils import normalize, text_hash

# Memuat data gabungan batch 3
file_path = "/content/drive/MyDrive/magang-ba-2026/data/processed/comments_batch3_merged.csv"
c = pd.read_csv(file_path)

# Menyesuaikan nama kolom teks agar sesuai dengan sintaks panduan (c.text_original)
# Jika kolom aslimu 'text_content', kita buat alias 'text_original' agar persis buku panduan
if 'text_content' in c.columns and 'text_original' not in c.columns:
    c['text_original'] = c['text_content']
```

```python
# 1. Hitung text_hash sesuai buku panduan
c["text_hash"] = c.text_original.map(text_hash)
```

```python
# 2. Cari duplikat pada pasangan (post_id, text_hash) dan duplikat comment_id
dup_post_text = c.duplicated(["post_id", "text_hash"], keep="first")
dup_comment_id = c.duplicated(["comment_id"], keep="first")
is_duplicate = dup_post_text | dup_comment_id
```

```python
# Cetak contoh duplikat persis seperti di panduan
dup = c[is_duplicate]
print(f"Jumlah duplikat ditemukan: {len(dup)}")
print("\nContoh data duplikat:")
print(dup[["comment_id", "post_id"]].head())

# Tandai duplikat sebagai eksklusi E2 di file processed (bukan di raw)
c["status_eksklusi"] = "Aktif"
c.loc[is_duplicate, "status_eksklusi"] = "E2"

# Simpan ke file processed
os.makedirs("/content/drive/MyDrive/magang-ba-2026/data/processed", exist_ok=True)
output_path = "/content/drive/MyDrive/magang-ba-2026/data/processed/comments_batch3_processed.csv"
c.to_csv(output_path, index=False)

print(f"\nFile processed berhasil disimpan di: {output_path}")
```

**Hitung jumlah data layak**

```python
import pandas as pd

# Memuat file processed yang sudah ditandai eksklusi E2
file_processed = "data/processed/comments_batch3_processed.csv"
df_proc = pd.read_csv(file_processed)

# Hitung rincian status
total_baris = len(df_proc)
jumlah_e2 = (df_proc["status_eksklusi"] == "E2").sum()
jumlah_layak = (df_proc["status_eksklusi"] == "Aktif").sum()

print(f"--- REKAPITULASI BATCH 3 ---")
print(f"Total baris awal: {total_baris}")
print(f"Total duplikat (E2): {jumlah_e2}")
print(f"Total data layak (Aktif): {jumlah_layak}")

# Target kumulatif (misalnya target minimal 1.000 data layak)
target_minimal = 1000
if jumlah_layak >= target_minimal:
    print(f"\nTarget minimal {target_minimal} data layak TERCAPAI!")
    kekurangan = 0
else:
    kekurangan = target_minimal - jumlah_layak
    print(f"\nBelum mencapai target. Kekurangan: {kekurangan} data untuk diganti di Hari 09.")
```
