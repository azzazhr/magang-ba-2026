**PLATFORM INSTAGRAM**

**Membaca data mentah dari file scrapping**

```python
import pandas as pd
import os

# 1. Memastikan folder data/processed tersedia
os.makedirs("data/processed", exist_ok=True)

# 2. Membaca data mentah dari folder data/raw/
df = pd.read_csv("/content/drive/MyDrive/magang-ba-2026/data/raw/sd_mv7xb0j12nq3spki17.csv").copy()
```

**Membersihkan data kosong**

```python
# 3. Bersihkan Data Kosong
df = df.dropna(subset=['comment'])
```

**Membuat ID Post**

```python
# 4. Mendapatkan URL unik untuk membuat ID Post
unique_urls = df['input_url'].unique()
post_id_map = {url: f"IG-P{str(i+1).zfill(3)}" for i, url in enumerate(unique_urls)}
df['post_id'] = df['input_url'].map(post_id_map)
```

**Menyiapkan kolom**

```python
# 5. Menyiapkan kolom yang dibutuhkan (tanpa username)
df['text_content'] = df['comment']
df['published_at'] = pd.to_datetime(df['comment_date']).dt.strftime('%Y-%m-%d')
df['eligibility_status'] = 'layak'
```

**Membuat komentar per post ID**

```python
# 6. Membuat ID Komentar berurutan per Post ID
df['comment_num'] = df.groupby('post_id').cumcount() + 1
df['comment_id'] = df['post_id'] + '-C' + df['comment_num'].astype(str).str.zfill(3)
```

```python
# 7. Membatasi maksimal 150 komentar per unggahan
df = df[df['comment_num'] <= 150]
```

```python
# 8. Memilih dan menyusun ulang kolom akhir
final_df = df[['comment_id', 'post_id', 'text_content', 'published_at', 'eligibility_status']]
```

**Menyimpan hasil akhir**

```python
# 9. Menyimpan hasil CSV bersih ke folder data/processed/
final_df.to_csv("data/processed/comments_instagram_b1.csv", index=False)

print("Proses selesai! File bersih berhasil disimpan di: data/processed/comments_instagram_b1.csv")
print(f"Total baris saat ini: {len(final_df)}\n")
print(final_df.head())
```

**SAMPLING QC**

```python
import pandas as pd
import os

# 1. Membaca data gabungan Batch 1 yang sudah bersih
file_merged = "data/processed/comments_batch1_merged.csv"

if os.path.exists(file_merged):
    df_merged = pd.read_csv(file_merged)
    total_data = len(df_merged)

    print(f"Total keseluruhan data Batch 1 saat ini: {total_data} baris")

    # 2. Menghitung jumlah sampel (5 sampel per 100 data, minimal 5 sampel)
    jumlah_sampel = max(5, int((total_data // 100) * 5))
    if jumlah_sampel > total_data:
        jumlah_sampel = total_data

    print(f"Mengambil {jumlah_sampel} sampel acak untuk pemeriksaan mutu...\n")

    # 3. Mengambil sampel secara acak (random sample)
    sampel_qc = df_merged.sample(n=jumlah_sampel, random_state=42)

    # 4. Menampilkan hasil sampel untuk dicocokkan dengan sumber asli
    print("=" * 60)
    print("SAMPEL ACAK UNTUK COCOK DATA DENGAN SUMBER ASLI:")
    print("=" * 60)
    for idx, row in sampel_qc.iterrows():
        print(f"• Comment ID  : {row['comment_id']}")
        print(f"  Post ID     : {row['post_id']}")
        print(f"  Tanggal     : {row['published_at']}")
        print(f"  Status      : {row['eligibility_status']}")

        # Mengatasi nilai kosong / NaN dengan mengubahnya menjadi string aman
        teks_aman = str(row['text_content']) if pd.notna(row['text_content']) else "[Komentar Kosong/Emoji]"
        print(f"  Teks Komentar: {teks_aman[:70]}...")
        print("-" * 60)

    # 5. Menyimpan hasil sampel QC ke folder docs/ untuk arsip audit
    os.makedirs("docs", exist_ok=True)
    sampel_qc.to_csv("docs/D06_sampel_qc_batch1.csv", index=False)
    print("\nDaftar sampel acak berhasil disimpan ke: docs/D06_sampel_qc_batch1.csv")
else:
    print("File gabungan comments_batch1_merged.csv belum ditemukan.")
    print("Tips: Pastikan kamu sudah menjalankan skrip penggabungan file IG dan TikTok sebelumnya.")
```
