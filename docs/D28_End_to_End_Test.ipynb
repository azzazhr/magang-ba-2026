**Membuat dan mengarahkan ke Basis Data Kosong Baru**

```python
import time
import sqlite3
from datetime import datetime

# Tentukan path database baru untuk Day 28
db_path = 'data/sentiment_d28.db'

# Pastikan folder 'data' ada
os.makedirs('data', exist_ok=True)

# Jika file database d28 sudah ada sebelumnya, hapus agar benar-benar bersih/kosong
if os.path.exists(db_path):
    os.remove(db_path)

# Inisialisasi tabel menggunakan fungsi dari modul src.db proyekmu
from src.db import init_db

# Jika init_db menggunakan path bawaan, kita bisa panggil atau buat tabel dasarnya langsung
conn = sqlite3.connect(db_path)
cursor = conn.cursor()

# Membuat tabel dasar yang dibutuhkan (posts, comments, predictions, verification_history)
cursor.execute('''
CREATE TABLE IF NOT EXISTS posts (
    post_id TEXT PRIMARY KEY,
    platform TEXT,
    published_at TEXT
)''')

cursor.execute('''
CREATE TABLE IF NOT EXISTS comments (
    comment_id TEXT PRIMARY KEY,
    post_id TEXT,
    platform TEXT,
    content_category TEXT,
    commented_at TEXT,
    text_original TEXT,
    text_content TEXT,
    context_text TEXT,
    is_eligible INTEGER,
    FOREIGN KEY(post_id) REFERENCES posts(post_id)
)''')

cursor.execute('''
CREATE TABLE IF NOT EXISTS predictions (
    prediction_id INTEGER PRIMARY KEY AUTOINCREMENT,
    comment_id TEXT,
    model_name TEXT,
    model_version TEXT,
    predicted_label TEXT,
    score REAL,
    score_type TEXT,
    predicted_at TEXT,
    run_id TEXT
)''')

cursor.execute('''
CREATE TABLE IF NOT EXISTS verification_history (
    verification_id INTEGER PRIMARY KEY AUTOINCREMENT,
    comment_id TEXT,
    prediction_id INTEGER,
    previous_label TEXT,
    new_label TEXT,
    reason TEXT,
    reviewer_id TEXT,
    verified_at TEXT
)''')

conn.commit()
conn.close()

print(f"Basis data kosong baru berhasil dibuat di: {db_path}")
```

```python
# Membaca isi file src/db.py saat ini dan mengganti path database-nya ke d28
db_file_path = 'src/db.py'

with open(db_file_path, 'r', encoding='utf-8') as f:
    content = f.read()

# Ganti path database lama dengan 'data/sentiment_d28.db'
# (Sesuaikan string di bawah ini dengan variabel path yang ada di file db.py milikmu)
if 'sentiment_d28.db' not in content:
    # Contoh umum penggantian path di db.py
    # Pastikan variabel penampung path di db.py-mu tertulis atau kita arahkan langsung
    updated_content = content.replace('sentiment.db', 'sentiment_d28.db').replace('sentiment_clean.db', 'sentiment_d28.db')

    with open(db_file_path, 'w', encoding='utf-8') as f:
        f.write(updated_content)
    print("Konfigurasi database di src/db.py berhasil diarahkan ke sentiment_d28.db!")
else:
    print("Konfigurasi database di src/db.py sudah menggunakan sentiment_d28.db.")
```
